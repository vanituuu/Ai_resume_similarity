```python
df = df.drop_duplicates(subset=["Role","Features"])
df = df.reset_index(drop=True)
```

```python
df.duplicated().sum()
```

```python
import re

def clean_text(text):

    text = str(text).lower()

    text = re.sub(r"[^a-zA-Z ]", " ", text)

    text = " ".join(text.split())

    return text
```

```python
df["clean_Feature"]= df["Features"].apply(clean_text)
```

```python
# pdf 
def extract_text_from_pdf(pdf_path):
    text = ""

    with pdfplumber.open(pdf_path) as pdf:

        for page in pdf.pages:

            page_text = page.extract_text()

            if page_text:
                text+=page_text+"\n"

    return text
```

```python
resume_text = extract_text_from_pdf("Vanita Resume.pdf")

```

```python
resume_text
```

```python
job_text = df["clean_Feature"]
```

```python
print(type(resume_text))
```

```python
print(type(job_text))
```

```python
all_text = [resume_text]+job_text.tolist()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf= TfidfVectorizer(stop_words="english")
vectors = tfidf.fit_transform(all_text)
```

```python
resume_vectors = vectors[0]

resume_vectors
```

```python
job_vectors = vectors[1:]

job_vectors
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(resume_vectors[0],job_vectors)[0]

df["similarity"] = similarity

top_jobs = (df.sort_values("similarity", ascending=False).drop_duplicates(subset=["Role"]).head(5))

```
